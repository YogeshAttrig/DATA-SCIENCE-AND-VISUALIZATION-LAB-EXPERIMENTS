```python
import pandas as pd
data ={
    "Name":["Tej", "Yash", "Yogesh","Yuvraj","Shubham"],
    "Age":[20,21,22,23,24],
    "Department":["CSE","It","CSE","ECE","IT"],
    "Marks":[85,90,91,88,77]
}
df= pd.DataFrame(data)
print(df)
```

```python
print (df.head())
print(df.tail())
print("shape:",df.shape)
print("columns:",df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print(df.dtypes)
print("\nInformation:")

print("\nDescriptive statistics:")
print(df.describe())
```

```python
df.to_csv("student_output.csv", index = False)
```
